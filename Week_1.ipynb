## Graphical Representation of Dataset

### Graph 1: Distribution of Target Variable(Cardiovascular Disease)
The graph shows the distribution of patients with and without cardiovascular disease.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sns.countplot(x='cardio', data=df, color='#F9A8D4')
plt.title("Distribution of Cardiovascular Disease")
plt.xlabel("Cardiovascular Disease (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

### Graph 2: Age Distribution
The histogram shows the distribution of patients' ages in the dataset.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age (Days)")
plt.ylabel("Frequency")
plt.show()
```

### Graph 3: Gender Distribution
This graph shows the number of male and female patients in the dataset.

```python
gender_counts = df['gender'].value_counts()
colors = ['#FDA4AF', '#C084FC']
plt.figure(figsize=(6,6))
plt.pie(gender_counts,
        labels=['Women','Men'],
        autopct='%1.1f%%',
        startangle=90,
        colors = colors
       )
plt.title("Gender Distribution")
plt.show()
```

### Graph 4: Cholesterol Level Distribution
The graph displays the frequency of each cholesterol level.

```python
plt.figure(figsize=(6,4))
sns.countplot(x='cholesterol', data=df)
plt.title("Cholesterol Level Distribution")
plt.xlabel("Cholesterol Level")
plt.ylabel("Count")
plt.show()
```

### Graph 5: Correlation Heatmap
- The heatmap represents the correlation between all numerical variables.
- Darker colors indicate stronger positive or negative relationships.
- Features with higher correlation to the target variable can be useful for prediction.

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
